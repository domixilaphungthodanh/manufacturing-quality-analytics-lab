## SQL mẫu → giải thích
Chưa kiểm tra không phải đạt. LEFT JOIN giữ đơn vị chưa có kiểm tra.

```python
read_sql('SELECT u.unit_code FROM units u WHERE NOT EXISTS(SELECT 1 FROM inspections i WHERE i.unit_id=u.unit_id AND i.inspection_no=1) LIMIT 20;')
```

## Thử thay đổi
So cấp dữ liệu, số dòng và mẫu số của truy vấn thứ hai.

```python
read_sql('SELECT u.unit_code,i.inspected_at FROM units u LEFT JOIN inspections i ON i.unit_id=u.unit_id AND i.inspection_no=1 ORDER BY u.unit_id LIMIT 20;')
```
